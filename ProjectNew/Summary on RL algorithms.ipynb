# Summary on "[Deep multiagent reinforcement learning: challenges and directions](https://arxiv.org/abs/2106.15691)".

## Article by Annie Wong, Thomas Bäck, Anna V. Kononova, Aske Plaat

## 1. Summary and Intoduction
The paper summarizes the representation problems in multi-agent environments. These problems are structured and presented in a simple way to facilitate future research. The key challenges discussed include centralized training and decentralized execution, opponent modeling, communication, efficient coordination, and reward shaping.

The paper covers basic concepts of reinforcement learning, explaining Markov Decision Processes (MDPs) in single-agent environments. The transition from single-agent to multi-agent reinforcement learning (MARL) introduces additional challenges, as the environment dynamics are determined by the joint actions of all agents. This results in non-stationarity, where each agent faces the "moving target" problem—constantly adapting to other agents' changing behaviors.

The curse of dimensionality is another significant challenge, as adding more agents drastically increases the state-action space. However, MARL also provides opportunities for agents to learn from one another and share experiences, which can accelerate the learning process.

Deep MARL is a rapidly expanding field capable of modeling many complex real-world problems. The authors propose a taxonomy based on the challenges inherent in multi-agent problem formulations and their solutions.

## 2. Single agent environments
The paper continues with an explanation of the basic Markov Decision Process (MDP), which constitutes a model for sequential decision-making under uncertainty, defining the interaction between a learning agent and its environment. An MDP is defined as a tuple: (S, A, P, R, $\gamma$), where S is set of states, A is set of actions, P is the transition probability function, R is the reward function and $\gamma \in [0,1]$ is discount coefficient defining the weight of future rewards: 
$$P(s', r \mid s, a) = P \{ S_t = s', R_t = r \mid S_{t-1} = s, A_{t-1} = a \}$$
This is the probability of transition to state $s'$ and receive a reward $r$ given that the sistem was in state $s$ and took an action $a$. Each state in an MDP satisfies the Markov Property, meaning that the future depends only on the current state, not on the history of earlier states and actions. Additionally, an MDP assumes full observability of the states and a stationary environment (i.e., the transition dynamics do not change over time).

Most of MDP solving algorithms fall into the following categories:
- value-based
- policy-based
- model-based

### Markov Game
Markov Game provide a theoretical framework to study multiple interacting agents in a fully observable
environment and can be applied to cooperative, collaborative and mixed settings. A Markov game is a collection of normal-form games (or matrix games) that the agents play repeatedly. Each state of the game can be viewed as a matrix representation with the payoffs for each joint action determined by the matrices.
