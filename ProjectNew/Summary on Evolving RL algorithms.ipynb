# Summary on "[Evolving Reinforcement Learning Algorithms](https://arxiv.org/abs/2101.03958)".
**John D. Co-Reyes, Yingjie Miao, Daiyi Peng, Esteban Real, Sergey Levine, Quoc V. Le, Honglak Lee, Aleksandra Faust**

## 1. Summary and Intoduction
The paper describes the process of "inventing" new ways of learning reinforcement learning, or so-called "meta-learning".
The problem to solve is the inability of AI to apply its learned knowledge to different environments. So the authors propose a method of teaching an AI algorithm to create its own RL algorithms. Their methods can both work from scratch—"re-discover" DQN, value-based algorithms—or bootstrap, starting from DQN and working towards the invention of unknown algorithms.

The purpose behind their work is that instead of humans designing new RL algorithms, AI can evolve better learning strategies automatically, find and improve RL strategies without human intervention, and avoid manual fine-tuning. So they are basically suggesting AI invent second-generation AI.

They are proposing a "universal language" and "optimization" of an outer loop that aims to maximize the reward of the inner loop, thus generalizing across many different and unseen environments. The language allows existing algorithms to be embedded into the starting graphs, leading to faster learning and interpretable algorithms. Their analysis of the meta-learned programs showed that the methods proposed automatically discovered algorithms that shared structures with RL innovations and empirically outperformed deep Q-learning methods.

## 3. Learning RL Algorithms
In the section, the authors stipulate the problem and their solution. The basics - we have a Q-function - state-action value function based algorithm and related descrete action-space. The policy is obtained using $\epsilon$-greedy strategy. The performance of an algorithm for a given environment
is summarized by the normalized average training return 
$$\frac{1}{M} \sum_{m=1}^{M} \frac{R_i - R_{\min}}{R_{\max} - R_{\min}}$$ where $\text{R}_{min}$ and $\text{R}_{max}$ are known ahead of time. The evaluation/output of the inner loop is then normalized and passed to the outer loop and it is: $$L^* = \arg\max_{L} \sum_{E} \text{Eval}(L, E)$$The goal of the meta-learnner is to find optimal loss function L(st, at, rt, st+1, θ, γ).

The authors propose a search language for reinforcement learning (RL) algorithms that is flexible enough to represent existing algorithms while allowing AI to discover new learning strategies.
Instead of just tweaking existing RL models, their approach learns new RL algorithms by optimizing policy updates (how an agent chooses actions) rather than focusing on exploration rewards.

## 4. Learner RL Algorithm results and analysis.

The link to their work can be found [here](https://github.com/google/brain_autorl/tree/main/evolving_rl) with additional details [here](https://sites.google.com/view/evolvingrl)

Then a description of the parameters of the expirement follow with the results, the learning convergence with the conclusions that the range of difficulty on the training environments is important for learning algorithms which can generalize and when compared with bootstrapping, the training performance does not saturate, leaving room for improvement. Bootstrapping from DQN significantly improved both the convergence and performance of the
meta-training, resulting in a 40% increase in final training performance.

They follow by showing interesting loss functions, $\text{DQN}_{CLIPPED}$ AND $\text{DQN}_{REG}$. Both methods have good generalization performance on the test environments.
$$L_{\text{DQNClipped}} = \max \left( Q(s_t, a_t), \delta^2 + Y_t \right) 
+ \max \left( Q(s_t, a_t) - Y_t, \gamma \left( \max_{a} Q_{\text{targ}}(s_{t+1}, a) \right)^2 \right)$$

The second learn loss is:
$$
L_{\text{DQNReg}} = 0.1 \cdot Q(s_t, a_t) + \delta^2
$$
DQNReg was trained from bootstrapping off DQN using three training environments. In comparison to DQNClipped, DQNReg directly regularizes the Q values with a weighted term that is always active. 

Both of these loss functions modify the original DQN loss function to regularize the Q-values to be lower in value. 

They established that while DQNReg is quite simple, it matches or outperforms the baselines on all training and test environments and solves the tasks when other methods fail to attain any reward.


## 5. Conclusions
This research demonstrates that AI can learn its own RL algorithms, reducing the need for human-designed RL strategies. While promising, the approach is currently limited to value-based RL and could be expanded to support more complex learning methods and diverse environments in the future.
